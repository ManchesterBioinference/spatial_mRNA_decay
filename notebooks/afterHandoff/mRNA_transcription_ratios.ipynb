```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# Read the transcription data
transcription_df = pd.read_csv('/mnt/mr01-home01/m65338lb/projects/spatial_mRNA_decay/results_1200/figures/intermediate/transcription/transcription_heatmap_stripe3_data.csv')
# Set yBin as index for both
transcription_df.set_index('yBin', inplace=True)

# Read the mRNA data
mrna_df = pd.read_csv('/mnt/mr01-home01/m65338lb/projects/spatial_mRNA_decay/results_1200/figures/intermediate/mrna/stripe3/e2_sass_formodel_heatmap_data.csv')
# Set yBin as index for both
mrna_df.set_index('yBin', inplace=True)
mrna_df.columns = transcription_df.columns  # Ensure columns match
mrna_df.index = transcription_df.index  # Ensure index matches


# Compute the ratio: transcription / mRNA
ratio_df = transcription_df.div(mrna_df)

# Plot the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(ratio_df, annot=True, cmap='viridis')
plt.title('Ratio of Transcription to mRNA')
plt.xlabel('Columns')
plt.ylabel('yBin')
plt.show()
```
